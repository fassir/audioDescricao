```python
import os
import time
from bot import ConversationBot
from audio_utils import record_audio
from IPython.display import Audio, display, clear_output

# Ensure API Key is set
# os.environ['OPENAI_API_KEY'] = 'your-key-here' # Uncomment if not using .env
```

```python
bot = ConversationBot()
```

```python
def interact():
    print("Recording in 3 seconds... Prepare to speak!")
    time.sleep(3)
    
    filename = "notebook_input.wav"
    record_audio(filename, duration=5) # 5 seconds fixed for notebook simplicity
    
    # Playback what was recorded
    display("You said (Audio):")
    display(Audio(filename))
    
    try:
        text = bot.transcribe_audio(filename)
        print(f"You said (Text): {text}")
        
        if text.strip():
            response = bot.get_gpt_response(text)
            print(f"Bot: {response}")
            
            out_file = bot.text_to_speech(response, output_file="notebook_response.mp3")
            display("Bot response (Audio):")
            display(Audio(out_file, autoplay=True))
    except Exception as e:
        print(f"Error: {e}")

interact()
```
